# Analisis Exploratorio de Datos (EDA)
Exploraremos los datos limpios para entender patrones de supervivencia de pasajeros del Titanic.

```python
import pandas as pd
import seaborn as sps
import matplotlib.pyplot as plt

df = pd.read_csv('data/titanic_clean.csv')
df.head(3)
```

# Supervivencia por género

```python
plt.figure(figsize=(5, 4))
sps.barplot(x='Sex', y='Survived', data=df, palette='viridis')
plt.title('Tasa de supervivencia por Género')
plt.show()
```

# Supervivencia por clase 

```python
plt.figure(figsize=(7, 4))
sps.barplot(x='Pclass', y='Survived', data=df, palette='magma')
plt.title('Tasa de supervivencia por Clase de Pasajero')
plt.show()
```

# Distribución por Edad

```python
plt.figure(figsize=(10, 6))
sps.histplot(data=df, x='Age', hue='Survived', kde=True, element='step')
plt.title('Distribución de Supervivencia por Edad')
```
