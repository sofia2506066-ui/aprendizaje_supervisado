# Ingeniería de Características
Vamos a preparar los datos para el modelo, codificando variables categoricas y generando algunas extras de ser necesario.


```python
import pandas as pd 

df = pd.read_csv('../data/titanic_clean.csv')
df.head(3)
```

# Codificación de Variables Categóricas 

```python
df["Sex"] = df["Sex"].map({"male": 0, "female": 1})

df = pd.get_dummies(df, columns=['Embarked'], prefix='Emb')

df.head()
```

```python
df.to_csv('../data/titanic_feature.csv', index=False)
print('Datos guardado en data/titanic_feature.csv')
```
